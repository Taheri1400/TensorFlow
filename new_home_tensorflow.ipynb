<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir">
<span style="color:red"><b>توجه:</b></span> پیش‌پردازش‌هایی که بر روی نمونه‌های اعتبارسنجی/آزمون اعمال می‌کنید باید طبق آمارها یا اطلاعاتی باشد که از روی نمونه‌های آموزشی به دست آمده است. به‌عنوان مثال اگر می‌خواهید مقادیر گم‌شده‌ی یک ویژگی را با میانگین آن ویژگی پُر کنید، باید از میانگینی که طبق نمونه‌های آموزشی محاسبه‌شده استفاده کنید. به همین دلیل، در کدگذاری نیز برای نمونه‌های اعتبارسنجی/آزمون باید از تابع <code dir=ltr>transform()</code> به‌جای <code dir=ltr>fit_transform()</code> استفاده کنید.
</font>
</p>

````python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

train['MaturitySize'] = le.fit_transform(train['MaturitySize'])
test['MaturitySize'] = le.transform(test['MaturitySize'])

# FurLength
train['FurLength'] = le.fit_transform(train['FurLength'])
test['FurLength'] = le.transform(test['FurLength'])

# Health
train['Health'] = le.fit_transform(train['Health'])
test['Health'] = le.transform(test['Health'])
````

<font face="vazir">
اما ستون‌های <code>Type</code>، <code>Breed1</code>، <code>Gender</code>، <code>Color1</code> ،<code>Color2</code> <code>Vaccinated</code> و <code>Sterilized</code>  از جنس متغیر‌های دسته‌ای اسمی هستند، به این معنی که ترتیب خاصی بین مقادیر گروه‌ها وجود ندارد. بنابراین برای این ویژگی‌ها نمی‌توانیم از <code>LabelEncoder</code> استفاده کنیم، زیرا این نوع داده‌ها از نظر ترتیبی تفاوتی با یک‌دیگر ندارند و اگر به‌سادگی به هر کدام از این ویژگی‌ها یک عدد اختصاص دهیم آن‌گاه به‌نوعی بین آن‌ها ترتیب قائل شدیم که نادرست است. برای حل این مسئله از روش کدگذاری دودویی استفاده می‌کنیم، به این معنی که به‌طور مثال به ازای ستون <code>color1</code> که ۷ مقدار یکتا <code dir=ltr>['Black', 'Brown', 'Cream', 'Gray', 'Golden', 'White', 'Yellow']</code> را دارد ۳ ستون اضافه می‌کنیم زیرا در هر ستون می‌توانیم مقدار ۱ و ۰ داشته باشیم و برای این‌که ۷ دسته را پوشش دهیم حداقل به ۳ ستون نیاز داریم (<code dir=ltr>7 < 3^2</code>). حال به هرکدام از این دسته‌ها عددی اختصاص می‌دهیم و ستون‌های مربوط به کد دودویی آن عدد را پر می‌کنیم.
<br>
    تابع <code>BinaryEncoder</code> از کتابخانه‌ی <code>category_encoders</code> این عملیات را انجام می‌دهد.
        <br>
    توجه داشته باشید در ستون <code>Gender</code> و <code>Type</code> به علت این‌که فقط ۲ دسته داریم می‌توانیم با تابع <code>LabelEncoder</code> تبدیل را انجام دهیم.
    </font>
</p>

````python
train['Type'] = le.fit_transform(train['Type'])
test['Type'] = le.transform(test['Type'])

train['Gender'] = le.fit_transform(train['Gender'])
test['Gender'] = le.transform(test['Gender'])
````

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">

نرمال سازی داده‌ها
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
 یکی از اعمال مهم پیش‌پردازش، تغییر مقیاس ویژگی‌ها به‌صورت توزیع نرمال است که اصطلاحاً به آن نرمال‌سازی (Normalization) می‌گوییم.
 نرمال‌سازی به کاهش تغییرات شدید در وزن‌ها کمک کرده و باعث سرعت بالاتر همگرایی مدل می‌شود.
برای انجام نرمال‌سازی در این تمرین باید اعداد هر ویژگی را طوری تغییر دهیم که میانگین آن‌ها <code>0</code> شود و واریانس <code>1</code> داشته باشند، این‌کار را می‌توانیم با استفاده از فرمول زیر انجام دهیم.
برای سری داده‌ی <code>X</code> (که در اینجا مقادیر هر ستون می‌باشد) به طوری‌که <code>X = [x_1, x_2, ..., x_n]</code>، با کم کردن میانگین از هر نمونه‌ی داده (<code>x_i</code>) و تقسیم آن بر انحراف معیار داده‌ها (سیگما) می‌توانیم سری داده‌ی نرمال‌شده را به‌دست آوریم.
</font>
</p>

$$ Z = \frac{x_i - \bar{x}}{\sigma} $$

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
دقت داشته باشید، از آنجا که در واقع ما فقط اطلاعات مربوط به مجموعه‌داده‌ی آموزش را در هنگام ساخت مدل داریم، برای نرمال‌سازی نمونه‌های آزمون نیز باید از میانگین و انحراف معیار نمونه‌های آموزش استفاده کنیم.
</font>
</p>

````python
for column in train.drop(columns=['Target']).columns:
  mean = train[column].mean() # TODO (Mean of train_data[column])
  std = train[column].std() # TODO (Standard Deviation of train_data[column])
  train[column] = (train[column]- mean)/std
  test[column]=  (test[column] - mean)/std

test.head()
````

    <ul dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir">
در نهایت از آن‌جایی که در کار با مدل نیاز هست جنس داده‌ها از نوع اعشار باشند، بنابراین جهت اطمینان داده‌ها را به <code>np.float32</code> تبدیل می‌کنیم.
</font>
</p>

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
   <code>__init__</code> تابع
</font>
</h3>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
در تابع <code>__init__(self)</code> مقداردهی اولیه برای لایه‌های شبکه را انجام دهید. توجه داشته باشید مقداردهی اولیه حتماْ به‌صورت <code>GlorotUniform</code> باشد. همچنین توجه داشته باشید که استفاده از لایه‌ی متراکم یا همان <code>Dense</code> مجاز نمی‌باشد، بنابراین در این تابع نیاز است، که فقط مقداردهی اولیه‌ی نورون‌ها با مقدار تصادفی و مقدار اولیه‌ی بایاس هر لایه (با مقدار <code>1</code>) را انجام دهید.

</p>

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
   <code>forward</code> تابع
</font>
</h3>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
در تابع <code>forward(self, x)</code>عملیات انتشار رو به جلو (forward propagation) را انجام خواهید داد.
    ورودی این تابع ورودی‌های لایه‌ی اولیه می‌باشد و خروجی این تابع خروجی مدل پس از رد شدن از تمامی‌ لایه‌هاست. توابع فعال‌ساز تمامی لایه‌ها به‌جز لایه‌ی خروجی تابع یکسوساز (ReLU) می‌باشد که می‌توانید از دستور <code>tf.nn.relu</code>
برای پیاده‌سازی آن استفاده کنید. تابع فعال‌ساز لایه‌ی خروجی را تابع سیگموید قرار دهید، که برای پیاده‌سازی آن می‌توانید از دستور <code>tf.nn.sigmoid</code> استفاده کنید.
</p>

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
   <code>predict</code> تابع
</font>
</h3>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
تابع <code>predict(self, x)</code> برای انتشار رو به جلوی نمونه‌های آزمون (یا همان پیش‌بینی نمونه‌های آزمون) پیاده‌سازی شده است، در این تابع قبل از فراخوانی تابع <code>forward(self, x)</code> ابتدا ورودی‌ها به‌صورت یک دسته‌ (batch) در می‌آیند. توجه داشته‌باشید نیازی به تغیر در این تابع نیست.
</p>

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
   <code>fit_one_batch</code> تابع
</font>
</h3>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
در تابع <code>fit_one_batch(self, inputs, targets, optimizer, loss_fn)</code> فرآیند آموزش مدل برای یک دسته را پیاده‌سازی کنید، این تابع علاوه‌بر ویژگی‌های ورودی و خروجی آن‌ها، تابع بهینه‌ساز و تابع هزینه را نیز دریافت می‌کند. <br>
 در این تابع  در یک گذر رو به جلو تابع <code>forward</code> فراخوانی می‌شود تا مقادیر خروجی به‌ازای نمونه‌های ورودی محاسبه شود. سپس با استفاده از تابع ضرر با مشاهده‌ی مقادیر خروجی شبکه و مقادیر حقیقی، مقدار ضرر را محاسبه کنید و با استفاده از تابع <code>tape.gradient</code> مشتق‌ها را محاسبه کنید. در نهایت با استفاده از مشتق‌های محاسبه‌شده و به‌کمک از تابع بهینه‌ساز یک مرحله آموزش را انجام دهید.

</p>

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
   <code>fit</code> تابع
</font>
</h3>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
در تابع <code>fit(self, dataset, valid_dataset, epochs, BATCH_SIZE, optimizer, loss_fn)</code> می‌خواهیم به تعداد دور‌های (epoch) دلخواه و با تعداد دسته (batch) دلخواه فرآیند آموزش را تکرار کنیم و در هر مرحله/دور از آموزش مدل، مقدار ضرر آموزش و اعتبارسنجی و همچنین مقدار f1 را روی داده‌های اعتبارسنجی نمایش دهیم، بنابراین این تابع علاوه بر مجموعه‌های آموزش و اعتبارسنجی، تعداد دور‌ها و اندازه‌ی دسته‌ها، و همچنین تابع ضرر و تابع بهینه‌ساز را نیز به‌عنوان ورودی‌ دریافت می‌کند (علت ورودی گرفتن توابع ضرر و بهینه‌ساز در این تابع این است که پس از ساخت یک شی از کلاس این تابع را برای آموزش فراخوانی می‌کنیم.).
به‌طور کلی می‌توان این تابع را به سه مرحله‌ تقسیم کرد.
</p>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
مقدار پیش‌فرض تابع ضرر را <code>tf.losses.BinaryCrossentropy</code>

 انتخاب کرده‌ایم.

در یک حلقه به اندازه تعداد دور‌های آموزش (epochs), به‌ازای تمامی دسته‌های مجموعه‌داده آموزش، با استفاده از تابع <code>fit_one_batch</code> فرایند آموزش را انجام دهید.

<h4 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
مرحله ۳: اعتبارسنجی</font>
</h4>
<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
در ابتدای تابع کدی وجود دارد که نمونه‌های اعتبارسنجی را تبدیل به یک دسته می‌کند. حال در این مرحله شما باید بعد از هر دور آموزش، با فراخوانی تابع <code>forward</code> خروجی مدل را برای نمونه‌های اعتبارسنجی به دست آورید. و سپس با استفاده از این پیش‌بینی‌ها مقدار ضرر اعتبارسنجی (val_loss) و مقدار <code>F1</code> برای این نمونه‌ها محاسبه شود.
</p>

````python
import tensorflow as tf
import numpy as np
from sklearn.metrics import f1_score
````

````python
class CustomModel(tf.Module):
    from sklearn.metrics import f1_score

    def __init__(self, input_features):

        initializer = tf.initializers.GlorotUniform() # TODO
        self.w1 = tf.Variable(initializer(shape=[input_features, 5000]), name = 'w1') #TODO
        self.bias1 = tf.Variable(tf.zeros([5000]), name = 'bias1') #TODO

        self.w2 = tf.Variable(initializer(shape=[5000, 1000]),name='w2') # TODO : don't forget to name variable name = 'w2'
        self.bias2 = tf.Variable(tf.ones([1000]), name='bias2') # TODO : don't forget to name variable name = 'bias1'

        self.w3 = tf.Variable(initializer(shape=[1000, 500]),name='w3') # TODO : don't forget to name variable name = 'w3'
        self.bias3 = tf.Variable(tf.ones([500]), name='bias3') # TODO : don't forget to name variable name = 'bias3'

        self.w_output = tf.Variable(initializer(shape=[500, 1]),name='_output') # TODO : don't forget to name variable name = 'w_output'
        self.bias_output = tf.Variable(tf.ones([1]), name='bias_output') # TODO : don't forget to name variable name = 'bias_output'


    def forward(self, x):

        self.first = tf.nn.relu(tf.matmul(x, self.w1) + self.bias1) # TODO : Output of 1st hidden layer, name it '1st'
        self.second = tf.nn.relu(tf.matmul(self.first, self.w2) + self.bias2) # TODO : Output of 2nd hidden layer, name it '2nd'
        self.third = tf.nn.relu(tf.matmul(self.second, self.w3) + self.bias3) # TODO : Output of 3rd hidden layer, name it '3nd'
        self.output = tf.nn.sigmoid(
            tf.matmul(self.third, self.w_output) + self.bias_output
        ) # TODO : Output of output layer or Output of model, name it 'output'

        return self.output

    def predict(self, x):
        batched_dataset = x.batch(len(x))
        X_batch = next(iter(batched_dataset))
        x = self.forward(X_batch)
        return x

    @tf.function
    def fit_one_batch(self, inputs, targets, optimizer, loss_fn):
        with tf.GradientTape() as tape:
            predictions = self.forward(inputs) #TODO
            loss = loss_fn(targets, predictions) #TODO

        gradients = tape.gradient(loss, self.trainable_variables) #TODO
        optimizer.apply_gradients(zip(gradients, self.trainable_variables)) #TODO

        return loss


    def fit(self, dataset, valid_dataset,  epochs, BATCH_SIZE,
            optimizer = tf.optimizers.Adam(), loss_fn = tf.losses.BinaryCrossentropy()):

        batched_validation_dataset = valid_dataset.batch(len(valid_dataset))
        X_valid, y_valid = next(iter(batched_validation_dataset))

        batched_dataset = dataset.batch(BATCH_SIZE) #TODO


        for epoch in range(epochs):
            for X_batch, y_batch in batched_dataset:
                loss = self.fit_one_batch(X_batch, y_batch, optimizer, loss_fn) # TODO

            val_predictions = self.forward(X_valid) # TODO : Predict Output of Validation Set
            val_loss = loss_fn(y_valid, val_predictions) # TODO : Validation Loss

            val_f1 = f1_score(y_valid.numpy().flatten(),(val_predictions.numpy().flatten() >= 0.5).astype(int),average='weighted')# TODO : validation f1 score (using scikit learn function)
            # NOTE : Please don't change << average = 'weighted' >> in the f1_score Function

            print(f'Epoch {epoch+1}, Loss: {loss}, Validation Loss: {val_loss.numpy()}, Validation F1 score: {val_f1}')
````
